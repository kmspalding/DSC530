## Permutation Testing

```python
import pandas as pd
import numpy as np

class DiffMeansPermute:

    def __init__(self, data, group_col1, group_col2, value_col):
        self.data = data
        self.group_col1 = group_col1  #Workout_Type
        self.group_col2 = group_col2  #Gender
        self.value_col = value_col      #Calories_Burned
        self.results = {}

    def TestStatistic(self, workout_type):
        #Calculates the difference in means for each gender within the specified workout type
        group1 = self.data[(self.data[self.group_col1] == workout_type) & (self.data[self.group_col2] == 'Male')][self.value_col]
        group2 = self.data[(self.data[self.group_col1] == workout_type) & (self.data[self.group_col2] == 'Female')][self.value_col]
        
        #Calculates absolute difference in means
        test_stat = abs(group1.mean() - group2.mean())
        return test_stat

    def MakeModel(self, workout_type):
        #Creates pooled dataset for permutation
        self.pool = self.data[self.value_col].values

    def RunModel(self, workout_type):
        #Split into two new groups
        np.random.shuffle(self.pool)
        n = len(self.data[(self.data[self.group_col1] == workout_type) & (self.data[self.group_col2] == 'Male')])
        data = self.pool[:n], self.pool[n:]
        return data

    def PValue(self, workout_type, iters=1000):
        self.MakeModel(workout_type)
        actual = self.TestStatistic(workout_type)
        count = 0
        for _ in range(iters):
            group1, group2 = self.RunModel(workout_type)
            perm_stat = abs(group1.mean() - group2.mean())
            if perm_stat >= actual:
                count += 1
        return actual, count / iters

    def run_tests(self):
        workout_types = self.data[self.group_col1].unique()
        for workout_type in workout_types:
            test_stat, p_value = self.PValue(workout_type)
            self.results[workout_type] = {'Test Statistic': test_stat, 'P-Value': p_value}

if __name__ == "__main__":
    exercise_dataset = pd.read_csv("C:/Users/k_spa/DSC530/gym_members_exercise_tracking.csv")
    test = DiffMeansPermute(exercise_dataset, 'Workout_Type', 'Gender', 'Calories_Burned')

    #Run tests for all four workout types
    test.run_tests()

    #The p-values were greater than the conventional significance level of 0.05, indicating that there were no statistically significant differences in calories burned between male and female participants for any of the workout types.
    for workout_type, result in test.results.items():
        print(f"Workout Type: {workout_type}, Actual Test Statistic: {result['Test Statistic']}, P-Value: {result['P-Value']}")
```

## Hypothesis Testing

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.formula.api import ols

class HypothesisTest(object):

    def __init__(self, data):
        self.data = data
        self.MakeModel()
        self.actual = self.TestStatistic(data)

    def PValue(self, iters=1000):
        self.test_stats = [self.TestStatistic(self.RunModel()) 
                           for _ in range(iters)]

        count = sum(1 for x in self.test_stats if x >= self.actual)
        return count / iters

    def TestStatistic(self, data):
        raise NotImplementedError("Subclasses should implement this!")

    def MakeModel(self):
        pass

    def RunModel(self):
        raise NotImplementedError("Subclasses should implement this!")

class ANOVATest(HypothesisTest):

    def TestStatistic(self, data):
        #Fits the ANOVA model
        model = ols('Calories_Burned ~ C(Workout_Type) * C(Gender)', data=data).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)
        
        #Returns the F-statistic 
        return anova_table['F']['C(Workout_Type):C(Gender)']

    def RunModel(self):
        #Creates a new dataset by shuffling calories_burned
        shuffled_data = self.data.copy()
        shuffled_data['Calories_Burned'] = np.random.permutation(shuffled_data['Calories_Burned'])
        return shuffled_data

if __name__ == "__main__":
    exercise_dataset = pd.read_csv("C:/Users/k_spa/DSC530/gym_members_exercise_tracking.csv")

    #Creates ANOVA test instance
    test = ANOVATest(exercise_dataset)

    #Calculates the p-value
    p_value = test.PValue(iters=1000)

    #Neither the type of workout nor the gender of the participants significantly affects the number of calories burned.
    print(f"Actual Test Statistic: {test.actual}")
    print(f"P-Value: {p_value}")
```
